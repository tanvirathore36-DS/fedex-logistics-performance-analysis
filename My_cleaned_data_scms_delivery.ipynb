```python
df = pd.read_csv('/content/drive/MyDrive/M2 - project (colab notebook and data set)/SCMS_Delivery_History_Dataset (1).csv')
```

```python
df.head()
```

```python
# Convert columns that should be dates
date_cols = [
    'PQ First Sent to Client Date',
    'PO Sent to Vendor Date',
    'Scheduled Delivery Date',
    'Delivered to Client Date',
    'Delivery Recorded Date'
]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce')
```

```python
# Convert numeric columns from string to numbers
df['Freight Cost (USD)'] = pd.to_numeric(df['Freight Cost (USD)'], errors='coerce')
df['Weight (Kilograms)'] = pd.to_numeric(df['Weight (Kilograms)'], errors='coerce')
df['Line Item Insurance (USD)'] = pd.to_numeric(df['Line Item Insurance (USD)'], errors='coerce')
```

```python
df.info()
```

```python
# Fill missing Shipment Mode and Dosage with labels
df['Shipment Mode'].fillna('Unknown', inplace=True)
df['Dosage'].fillna('Not Available', inplace=True)

# Fill missing numeric values with 0
df.fillna({
    'Line Item Insurance (USD)': 0,
    'Weight (Kilograms)': 0,
    'Freight Cost (USD)': 0
}, inplace=True)
```

```python
df.isnull().sum()
```

```python
# Calculate delay in days
df['Delay_Days'] = (df['Delivered to Client Date'] - df['Scheduled Delivery Date']).dt.days

# Create On-Time / Delayed flag
df['OnTime'] = df['Delay_Days'].apply(lambda x: 'On Time' if x <= 0 else 'Delayed')
```

```python
df[['Scheduled Delivery Date', 'Delivered to Client Date', 'Delay_Days', 'OnTime']].head(10)
```

```python
# Export the cleaned dataset to a CSV file
df.to_csv("FedEx_Cleaned.csv", index=False)

print("✅ Cleaned dataset exported successfully! File name: FedEx_Cleaned.csv")
```
